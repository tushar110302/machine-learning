```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("wine_data.csv", header=None, usecols=[0, 1, 2])
df.columns=['Class Label', 'Alcohol', 'Malic Acid']
df.head()
```

```python
sns.kdeplot(x=df["Alcohol"])
```

```python
sns.kdeplot(x=df["Malic Acid"])
```

```python
sns.scatterplot(x=df["Alcohol"], y=df["Malic Acid"], hue=df["Class Label"])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop("Class Label", axis=1), df["Class Label"], test_size=0.2)
X_train.shape, X_test.shape
```

```python
y_train
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```
